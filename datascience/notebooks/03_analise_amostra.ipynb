# 🔎 Análise Amostra

## Origem dos Dados

Na fase da coleta de dados foi gerado o arquivo artigos_en.csv com 2.010 artigos em inglês. A partir desses dados foi gerado o arquivo artigos_en_amostra.csv com 201 artigos sorteados aleatoriamente (cerca de 10% por subcategoria).

A amostra gerada foi validada manualmente pelo time. Cada artigo foi lido e avaliado considerando as seguintes notas: 

0 = categoria incorreta

1 = categoria parcialmente adequada

2 = categoria correta

As notas foram registradas em planilha do Google Sheets e fornecidas para categoria, subcategoria e tags do texto. Esses dados foram utilizados para definir a qualidade dos dados para o projeto.

## Avaliação da Amostra

Cada registro da amostra foi avaliado individualmente pelos membros da equipe, para facilitar a validação manual, a amostra foi organizada em uma planilha compartilhada, contendo as principais informações de cada artigo, como id, categoria, subcategoria, título, texto, tags, url e Obs quando necessário. Cada integrante da equipe identificava seu nome na linha do registro que escolheu validar e a partir da URL realizava a leitura do artigo. Após essa análise, cada integrante registrava as notas de Texto x Categoria.
Avaliação das tags e subcategorias 

As subcategorias e as tags também foram avaliadas individualmente em relação à categoria atribuída. As subcategorias são utilizadas como parâmetros de busca na API do DEV.to, que utiliza as tags atribuídas pelos autores para localizar e retornar os artigos. Dessa forma, a avaliação humana permitiu verificar a qualidade dos artigos recuperados por cada busca: caso uma tag apresentasse baixo desempenho ou retornasse conteúdos pouco relacionados à categoria esperada, sua busca poderia ser descartada e substituída por outra subcategoria/tag, buscando recuperar textos mais adequados à categoria utilizada no treinamento.
Métricas de avaliação 

Para medir a qualidade das categorias, foram calculados os percentuais de classificações corretas e de classificações parcialmente adequadas:

Nota 2: quantidade de registros com nota 2 / quantidade total de registros da categoria × 100
Notas 1 + 2: quantidade de registros com nota diferente de 0 / quantidade total de registros da categoria × 100

**Resultados** 

Os resultados de Texto x Categoria foram comparados entre as seis categorias:

< categoria> (Nota 2 / Nota 1+2)

Backend (78,79% / 93,94%)
Cloud (48,57% / 54,29%)
DataScience (54,84% / 70,97%)
Database (68,75% / 75,00%)
Frontend (68,57% / 77,14%)
Security (85,71% / 97,14%)

No conjunto da amostra, 67,66% dos registros receberam nota 2 e 78,11% receberam nota 1 ou 2.

**Análise das subcategorias** 

A decisão de manter ou eliminar subcategorias considerou tanto os indicadores quantitativos quanto a análise qualitativa das observações registradas durante a validação. Foram identificados, por exemplo:

Cloud/OCI (33,33% de nota 2)
DataScience/AI (50%)
DataScience/Deep Learning (0%)
Frontend/TypeScript (0%)
Database/MongoDB (60%)

Entretanto, uma baixa porcentagem não foi considerada, isoladamente, motivo para eliminação, pois algumas dessas subcategorias ainda contribuem para trazer textos relevantes e representativos para o treinamento do modelo. No caso de IA, por exemplo, foi observado que muitos dos conteúdos retornados tratavam de Inteligência Artificial de forma mais independente de DataScience, portanto, tanto o desempenho percentual quanto a relevância dos registros para a categoria.

**Tratamento dos conflitos**

Durante a análise para a limpeza dos dados foram identificados 71 casos em que o mesmo ID e texto estavam associados a categorias diferentes, decorrentes principalmente de artigos que possuíam múltiplas tags. Para resolver esses conflitos de forma padronizada, foram estabelecidas regras hierárquicas:

* Maior nota de desempenho da subcategoria
* Em caso de empate, menor quantidade de registros associados à subcategoria
* Persistindo o empate, menor quantidade de registros associados à categoria

Essas regras permitiram selecionar a classificação com maior evidência de coerência na validação, reduzindo duplicidades e evitando decisões subjetivas.

**Resultado final da amostra**

A análise identificou subcategorias com menor aderência entre os artigos recuperados e a categoria atribuída. Como resultado, foram eliminadas as subcategorias OCI (Cloud), AI (DataScience) e DeepLearning (DataScience), considerando seu desempenho na validação e a qualidade dos conteúdos recuperados. 

Após esse filtro, a amostra foi considerada adequada para orientar a seleção das subcategorias utilizadas na coleta, permitindo reduzir buscas com menor aderência e priorizar aquelas com maior capacidade de recuperar conteúdos coerentes com suas respectivas categorias.
Após o descarte das subcategorias com menor desempenho na validação da amostra, restaram 1.851 registros, utilizados como entrada para a etapa de limpeza dos dados.
